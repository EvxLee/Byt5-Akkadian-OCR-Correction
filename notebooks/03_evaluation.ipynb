# 02 — Evaluation
Computes exact match, CER, chrF++, and BLEU for:
1. **Baseline:** noisy input vs gold (tells you what zero-shot model improvement means)
2. **Model:** fine-tuned ByT5 predictions vs gold

**Input:** `results/test_pairs.jsonl` (noisy, gold), `results/model_predictions.txt`  
**Output:** printed report

```python
import sys
sys.path.append('..')

import json
from pathlib import Path
from src.metrics.evaluate import full_report
```

```python
TEST_PAIRS = Path('../results/test_pairs.jsonl')
PREDICTIONS = Path('../results/model_predictions.txt')

pairs = [json.loads(l) for l in TEST_PAIRS.read_text().splitlines()]
noisy_inputs = [p['noisy'] for p in pairs]
gold_refs    = [p['gold']  for p in pairs]

model_preds = PREDICTIONS.read_text().splitlines()
assert len(model_preds) == len(gold_refs), "Prediction count mismatch"
```

```python
baseline = full_report(noisy_inputs, gold_refs, label='baseline (noisy input)')
model    = full_report(model_preds,  gold_refs, label='byt5-small fine-tuned')

for report in [baseline, model]:
    print(f"\n=== {report['label']} ===")
    print(f"  Exact match : {report['exact_match']:.3f}")
    print(f"  CER         : {report['cer']:.3f}  (lower is better)")
    print(f"  chrF++      : {report['chrf']:.2f}")
    print(f"  BLEU        : {report['bleu']:.2f}")
```
